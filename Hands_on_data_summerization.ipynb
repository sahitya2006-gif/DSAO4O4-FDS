```python
#2 Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset
df = pd.read_csv("/content/StudentsPerformance.csv")

# Display first 5 rows
print("First 5 Records:")
print(df.head())

# Display column names
print("\nColumn Names:")
print(df.columns)

# Create Grade Column
def get_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['math score'].apply(get_grade)

# Frequency Distribution
grade_count = df['Grade'].value_counts().sort_index()

print("\nFrequency Distribution of Grades:")
print(grade_count)

# Plot Bar Chart
plt.figure(figsize=(8,5))

plt.bar(
    grade_count.index,
    grade_count.values,
    color='skyblue',
    edgecolor='black',
    width=0.6
)

plt.title("Frequency Distribution of Students' Grades", fontsize=14)
plt.xlabel("Grades", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
#8 Titanic dataset and perform Exploratory Data Analysis
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("/content/Titanic_Cleaned.csv")
# ----------------------------------------------------
# 1. Display Dataset Information
# ----------------------------------------------------
print("First 5 Rows")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nShape of Dataset")
print(df.shape)

print("\nColumn Names")
print(df.columns)

# ----------------------------------------------------
# 2. Find Missing Values
# ----------------------------------------------------
print("\nMissing Values")
print(df.isnull().sum())

# ----------------------------------------------------
# 3. Descriptive Statistics
# ----------------------------------------------------
print("\nDescriptive Statistics")
print(df.describe())

# ----------------------------------------------------
# 4. Plot Histograms
# ----------------------------------------------------
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns].hist(figsize=(15,10), bins=15, edgecolor='black')
plt.suptitle("Histograms of Numerical Columns")
plt.tight_layout()
plt.show()

# ----------------------------------------------------
# 5. Plot Box Plots
# ----------------------------------------------------
plt.figure(figsize=(15,10))

for i, col in enumerate(numeric_columns, 1):
    plt.subplot(3,3,i)
    sns.boxplot(y=df[col], color="skyblue")
    plt.title(col)

plt.tight_layout()
plt.show()

# ----------------------------------------------------
# 6. Detect Outliers Using IQR
# ----------------------------------------------------
print("\nOutliers Detected")

outlier_columns = []

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    if len(outliers) > 0:
        outlier_columns.append(col)

    print(f"{col}: {len(outliers)} outliers")

# ----------------------------------------------------
# 7. Remove Outliers
# ----------------------------------------------------
cleaned_df = df.copy()

for col in numeric_columns:

    Q1 = cleaned_df[col].quantile(0.25)
    Q3 = cleaned_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    cleaned_df = cleaned_df[
        (cleaned_df[col] >= lower) &
        (cleaned_df[col] <= upper)
    ]

print("\nOriginal Shape :", df.shape)
print("Cleaned Shape :", cleaned_df.shape)

# ----------------------------------------------------
# 8. Save Cleaned Dataset
# ----------------------------------------------------
cleaned_df.to_csv("Titanic_Cleaned.csv", index=False)

print("\nCleaned dataset saved as Titanic_Cleaned.csv")
```
